```python
import pandas as pd
from getpass import getpass
from sqlalchemy import create_engine

clean = pd.read_csv("electronics_clean.csv")

password = getpass("MySQL password: ")
engine = create_engine(f"mysql+pymysql://root:{password}@localhost:3306/costco_electronics")

clean.to_sql("products", engine, if_exists="replace", index=False)
print("Loaded", len(clean), "rows")
```

```python
clean["title"] = clean["title"].str.encode("ascii", "ignore").str.decode("ascii").str.strip()
```
